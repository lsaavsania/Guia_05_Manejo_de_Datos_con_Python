### CREACIÓN DE TABLAS RESUMEN / FRECUENCIA  -- INTERPRETACIÓN DE LOS RESULTADOS

```python
# 24. Mostrar una tabla de frecuencias del "riesgo" de los clientes.

df["riesgo"].value_counts()
```

```python
# Interpretación
# La tabla muestra la distribución de los clientes según su nivel de riesgo. 
# De un total de 80 clientes, 39 presentan riesgo medio, 28 riesgo bajo y 13 riesgo alto.
# Por lo tanto, el nivel de riesgo más frecuente es el medio, mientras que el riesgo alto es el menos frecuente.
```

```python
# 25. Mostrar una tabla de frecuencias del "plazo_meses" de los clientes.

df["plazo_meses"].value_counts()
```

```python
# Interpretación
# La tabla muestra la distribución de los clientes según el plazo del crédito en meses. 
# El plazo más frecuente es de 48 meses, con 17 clientes, seguido de 6 y 36 meses, con 16 clientes cada uno.
# El plazo menos frecuente es de 18 meses, con 8 clientes. 
# En general, se observa que los clientes presentan distintos plazos de crédito, con una mayor frecuencia en los 48 meses.
```

```python
# 26. Mostrar una tabla resumen del promedio del "ingreso_mensual", del "monto_solicitado" y de la "deuda_mensual" 
# agrupado por "riesgo".

df.groupby("riesgo")[["ingreso_mensual","monto_solicitado","deuda_mensual" ]].mean()
```

```python
# Interpretación
# La tabla compara los promedios de ingreso mensual, monto solicitado y deuda mensual según el nivel de riesgo.
# Los clientes de riesgo alto presentan el menor ingreso mensual promedio (1908.33) y 
# la mayor deuda mensual promedio (2416.67). En cambio, 
# los clientes de riesgo bajo tienen el mayor ingreso mensual promedio (5119.23) y la menor deuda mensual promedio (743.48).
# Esto muestra que, en estos datos, un mayor nivel de riesgo está asociado con menores ingresos y mayores deudas mensuales.
```

```python
# 27. Mostrar una tabla resumen del máximo de "monto_solicitado" y "ingreso_mensual" agrupado por "riesgo".

df.groupby("riesgo")[["monto_solicitado","ingreso_mensual"]].max()
```

```python
# Interpretación
# La tabla muestra los valores máximos del monto solicitado y del ingreso mensual para cada nivel de riesgo.
# Los clientes de riesgo alto alcanzan un monto solicitado máximo de 20 000 y un ingreso mensual máximo de 3 000. 
# En riesgo medio, el monto máximo también es 20 000, pero el ingreso máximo llega a 8 000. 
# En riesgo bajo, el monto máximo solicitado es 16 000 y el ingreso máximo es 8 000.
# Se observa que el grupo de riesgo alto presenta un ingreso máximo considerablemente menor que los otros niveles de riesgo, 
# aunque puede solicitar montos de crédito elevados.
```

```python
# 28. Mostrar una tabla cruzada de frecuencias entre "riesgo" y "plazo_meses".Mostrar una tabla cruzada de frecuencias 
# entre "riesgo" y "atrasos_ultimos_12_meses".

pd.crosstab(df["riesgo"],df["atrasos_ultimos_12_meses"])
```

```python
# Interpretación
# La tabla muestra la cantidad de clientes según su nivel de riesgo y el número de atrasos en los últimos 12 meses.
# Se observa que los clientes de riesgo bajo y medio se concentran principalmente entre 0 y 1 atraso. 
# En cambio, en el grupo de riesgo alto predominan los clientes que presentan entre 1 y 3 atrasos.
# En general, los datos sugieren que una mayor cantidad de atrasos está asociada con un mayor nivel de riesgo crediticio.
```
